```python
import numpy as np
import pandas as pd
```

```python
dataset = pd.read_csv("Wine.csv")
X = dataset.iloc[:,0:13].values
y = dataset.iloc[:,13].values
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.decomposition import KernelPCA
kpca = KernelPCA(n_components=2,kernel='rbf',gamma=15)
X_train = kpca.fit_transform(X_train)
X_test = kpca.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
log=LogisticRegression(random_state=0,max_iter=1000)
log.fit(X_train,y_train)
log_acc=accuracy_score(y_test,log.predict(X_test))
```

```python
svml=SVC(kernel='linear',random_state=0)
svml.fit(X_train,y_train)
svml_acc=accuracy_score(y_test,svml.predict(X_test))
```

```python
svmnl=SVC(kernel='rbf',random_state=0)
svmnl.fit(X_train,y_train)
svmnl_acc=accuracy_score(y_test,svmnl.predict(X_test))
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train,y_train)
knn_acc=accuracy_score(y_test,knn.predict(X_test))
```

```python
nb=GaussianNB()
nb.fit(X_train,y_train)
nb_acc=accuracy_score(y_test,nb.predict(X_test))
```

```python
dt=DecisionTreeClassifier(random_state=0)
dt.fit(X_train,y_train)
dt_acc=accuracy_score(y_test,dt.predict(X_test))
```

```python
rf=RandomForestClassifier(n_estimators=10,random_state=0)
rf.fit(X_train,y_train)
rf_acc=accuracy_score(y_test,rf.predict(X_test))
```

```python
result=pd.DataFrame({
    "Accuracy":[
        log_acc,
        svml_acc,
        svmnl_acc,
        knn_acc,
        nb_acc,
        dt_acc,
        rf_acc
    ]
},
index=[
    "Logistic",
    "SVM Linear",
    "SVM Non Linear",
    "KNN",
    "Naive Bayes",
    "Decision Tree",
    "Random Forest"
])
```
